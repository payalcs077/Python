```python
import time
from datetime import datetime
a = time.time()
lastlogging = datetime.fromtimestamp(a)
print(lastlogging)
def update_last_login():
    a = time.time()
    global lastlogging
    lastlogging = datetime.fromtimestamp(a)
    return lastlogging
last_logging = update_last_login()
print("Last login time:", last_logging)
```
