```python
import asyncio

async def fetch_data(name,delay):
    print(f"fetching name {name}")
    await asyncio.sleep(delay)
    print(f" {name} fetched")

async def main():
    await asyncio.gather(
        fetch_data('Users',2),
        fetch_data('orders',3),
        fetch_data('product',1)
    )
# asyncio.run(main())
await main()
```

```python
class Student:
    def __init__(self, name,marks,rollno):
        self.name = name
        self.marks = marks
        self.rollno = rollno

    def add_marks(self,marks):
        self.marks += marks
    def isPass(self):
        if self.marks >=40:
            return True
    def display(self):
        print(f"Student {self.name} bearing rollno {self.rollno} passed the exam with marks {self.marks}")
student1 = Student("Payal",100,1)
student1.add_marks(50)
student1.display()
```
