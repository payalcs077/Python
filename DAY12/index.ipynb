```python
class Student:
    def __init__(self, name, total_classes):
        self.name = name
        self.total_classes = total_classes
        self.__attendance = 0   # private attribute

    def mark_present(self):
        self.__attendance += 1

    def mark_absent(self):
        pass  # attendance does not increase

    def get_attendance(self):
        return self.__attendance

    def is_eligible(self):
        attendance_percentage = (self.__attendance / self.total_classes) * 100
        return attendance_percentage >= 75
```

```python
s1 = Student("Payal", 10)

for _ in range(85):
    s1.mark_present()

print("Attendance:", s1.get_attendance())
print("Eligible:", s1.is_eligible())
```

```python
#the method which access the private attribute is called as Getter
#the method which updates the value of the private attribute is called as Setter
```

```python
class BankAccount:
    def __init__(self, name, balance=0):
        self.name = name
        self.__balance = balance  

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance.")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    def _display(self):
        print(f"Account Holder: {self.name}")
        print(f"Current Balance: ₹{self.__balance}")
```

```python
acc = BankAccount("Payal", 1000)
acc.deposit(500)
acc.withdraw(300)
acc.display()
```

```python
from abc import ABC, abstractmethod

class BankMainApp(ABC):

    @abstractmethod
    def api_call(self):
        pass

    @abstractmethod
    def db_call(self):
        pass

class WebApp(BankMainApp):

    def api_call(self):
        print("Secure API call from Web Application")

    def db_call(self):
        print("Secure Database call from Web Application")

app = WebApp()
app.api_call()
app.db_call()
```

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        print("Baby I am praying on you tonight, just like animals... just like animals... woooo ohhhh")

class Dog(Animal):
    def sound(self):
        print("Meowww")

class Cat(Animal):
    def sound(self):
        print("Bowwww")

    
```

```python
Dog().sound()
Cat().sound()
```

```python
class SecureLocker():
    def __init__(self,pin,is_locked):
        self.__pin = pin
        self.__is_locked = is_locked
    def __verify_pin(self,pin):
        if self.__pin == pin:
            return True
        else:
            return False
    def unlock(self, pin):
        if self.__verify_pin(pin):
            self.__is_locked = False
            print("Locker Unlocked")
        else:
            print("Wrong pin ")
    def lock(self):
        self.__is_locked = True
        print("Locker Locked")
    
    def __str__(self):
        return f"The staus of the locker is => Locked: {self.__is_locked}"
```

```python
l = SecureLocker(1000,True)
l.lock()
print(l)
```
